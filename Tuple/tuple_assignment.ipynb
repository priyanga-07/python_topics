```python
### Assignment 15: Tuple Functions

# Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.
def tup_min(numb):
    return min(numb)

def tup_max(numb):
    return max(numb)

def tup_sum(numb):
    return sum(numb)

tops=(1,2,3,4,5)
result_min=tup_min(tops)
result_max=tup_max(tops)
result_sum=tup_sum(tops)
print(f"min_result:{result_min}")
print(f"max_result:{result_max}")
print(f"sum_result:{result_sum}")
```
