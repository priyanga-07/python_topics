```python
### Assignment 2: Array Indexing and Slicing

# 1. Create a NumPy array of shape (6, 6) with values from 1 to 36.
# Extract the sub-array consisting of the 3rd to 5th rows and 2nd to 4th columns.

arr1=np.random.randint(1, 37, size=(6,6))
print(arr1)
print(arr1.shape)
print(arr1[2:5,2:4])
```

```python
# 2. Create a NumPy array of shape (5, 5) with random integers. Extract the elements on the border.

arr2=np.random.randint(1,26,size=(5,5))
print(arr2)

border = np.concatenate([
    arr2[0, :],      # top row
    arr2[-1, :],     # bottom row
    arr2[1:-1, 0],   # left column (excluding corners)
    arr2[1:-1, -1]   # right column (excluding corners)
])

print(border)
```

```python
### Assignment 3: Array Operations

# 1. Create two NumPy arrays of shape (3, 4) filled with random integers.
# Perform element-wise addition, subtraction, multiplication, and division.

arr3=np.random.randint(1,12,size=(3,4))
arr4=np.random.randint(1,12,size=(3,4))
print(f"array 1:{arr3}")
print(f"array 2 :{arr4}")

print(f"addition {arr3+arr4}")
print(f"subtraction {arr3-arr4}")
print(f"multiplication {arr3*arr4}")
print(f"division {arr3*arr4}")
```

```python
# 2. Create a NumPy array of shape (4, 4) with values from 1 to 16. Compute the row-wise and column-wise sum.

arr5=np.random.randint(1,16,size=(4,4))
print(arr5)

# Row-wise sum
row_sum = np.sum(arr5, axis=1)

# Column-wise sum
col_sum = np.sum(arr5, axis=0)

print("Row-wise sum:")
print(row_sum)

print("\nColumn-wise sum:")
print(col_sum)
```

```python
### Assignment 4: Statistical Operations

# 1. Create a NumPy array of shape (5, 5) filled with random integers.
# Compute the mean, median, standard deviation, and variance of the array.

arr6=np.random.randint(1,26,size=(5,5))
print(arr6)
print(f"mean {np.mean(arr6)}")
print(f"std {np.std(arr6)}")
print(f"median {np.median(arr6)}")
print(f"variance {np.var(arr6)}")
```

```python
# 2. Create a NumPy array of shape (3, 3) with values from 1 to 9.
# Normalize the array (i.e., scale the values to have a mean of 0 and a standard deviation of 1).

arr7=np.random.randint(1,10,size=(3,3))
print(arr7)
nor_dev=(arr7-np.mean(arr7)/np.std(arr7))
print(f"normal deviation: {nor_dev}")
```

```python

### Assignment 5: Broadcasting

# 1. Create a NumPy array of shape (3, 3) filled with random integers.
# Add a 1D array of shape (3,) to each row of the 2D array using broadcasting.
# Create a 3x3 array with random integers
arr_2d = np.random.randint(1, 10, size=(3, 3))

# Create a 1D array of shape (3,)
arr_1d = np.array([1, 2, 3])

# Add the 1D array to each row using broadcasting
result = arr_2d + arr_1d

print("Original 2D Array:")
print(arr_2d)

print("\n1D Array:")
print(arr_1d)

print("\nResult:")
print(result)
```

```python
# 2. Create a NumPy array of shape (4, 4) filled with random integers.
# Subtract a 1D array of shape (4,) from each column of the 2D array using broadcasting.

arr1_2d=np.random.randint(1, 17, size=(4,4))
print(f"array: {arr1_2d}")
arr2_2d=np.array([1,2,3,4])
print(f"array: {arr2_2d}")
sub=arr1_2d-arr2_2d
print(f"sub: {sub}")
```
