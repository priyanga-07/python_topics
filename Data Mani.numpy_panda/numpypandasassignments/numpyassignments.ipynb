```python
### Assignment 6: Linear Algebra

# 1. Create a NumPy array of shape (3, 3) representing a matrix. Compute its determinant, inverse, and eigenvalues.
import numpy as np
arr_linear=np.random.randint(1, 10, size=(3, 3))
print(f"array: {arr_linear}")
determinant=np.linalg.det(arr_linear)
print(f"determinant: {determinant}")
inverse=np.linalg.inv(arr_linear)
print(f"inverse: {inverse}")
eigenvalues=np.linalg.eig(arr_linear)
print(f"eigenvalues: {eigenvalues}")
```

```python
### Assignment 9: Structured Arrays

# 1. Create a structured array with fields 'name' (string), 'age' (integer), and 'weight' (float).
# Add some data and sort the array by age.
import numpy as np

# Define the structured data type
person_dtype = [('name', 'U20'), ('age', 'i4'), ('weight', 'f4')]

# Create the structured array
people = np.array([
    ('Alice', 25, 55.5),
    ('Bob', 30, 70.2),
    ('Charlie', 22, 68.0),
    ('David', 28, 75.3)
], dtype=person_dtype)

print("Original Array:")
print(people)

# Sort by age
sorted_people = np.sort(people, order='age')

print("\nSorted by Age:")
print(sorted_people)
```

```python
# 2. Create a structured array with fields 'x' and 'y' (both integers). Add some data and compute the Euclidean distance between each pair of points.

str_arr=[('x','i4'),('y','i4')]
field_arr=np.array([(10,20),
                    (20,30),
                    (30,40)],dtype=str_arr)
print(f"field_arr: {field_arr}")

# Compute distances between each pair of points
n = len(field_arr)

for i in range(n):
    for j in range(i + 1, n):
        dx = field_arr[j]['x'] - field_arr[i]['x']
        dy = field_arr[j]['y'] - field_arr[i]['y']

        distance = np.sqrt(dx**2 + dy**2)

        print(f"Distance between Point {i} and Point {j}: {distance:.2f}")
```

```python
### Assignment 10: Masked Arrays

# 1. Create a masked array of shape (4, 4) with random integers and mask the elements greater than 10. Compute the sum of the unmasked elements.

mask_arr=np.random.randint(1,17,size=(4,4))
print(f"mask_arr: {mask_arr}")
mas_gr=np.ma.masked_greater(mask_arr, 10)
print(f"mass_gr: {mas_gr}")
mas_sum=mas_gr.sum()
print(f"mass_sum: {mas_sum}")
```

```python
import numpy as np

# Create a 3x3 array with random integers from 1 to 20
arr = np.random.randint(1, 21, size=(3, 3))

print("Original Array:")
print(arr)

# Mask the diagonal elements
masked_arr = np.ma.masked_array(arr, mask=np.eye(3, dtype=bool))

print("\nMasked Array:")
print(masked_arr)

# Mean of unmasked elements
mean_value = masked_arr.mean()

print("\nMean of unmasked elements:")
print(mean_value)

# Replace masked elements with the mean
filled_arr = masked_arr.filled(mean_value)

print("\nArray after replacing masked elements:")
print(filled_arr)
```
