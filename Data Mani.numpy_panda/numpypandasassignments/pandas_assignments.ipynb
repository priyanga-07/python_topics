```python
### Assignment 4: Data Aggregation

# 1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'.
# Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers.
# Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

# with 2 columns: 'Category' and 'Value'.
import pandas as pd
import numpy as np
data6={
    'Category':['A','B','C'],
    # 'Values':[10,20,30]
    'values':np.random.randint(1,10,size=(3))
}
# categories ('A', 'B', 'C') and the 'Value' column with random integers.
df6=pd.DataFrame(data6,columns=['Category','values'])
print(f"original dataframe")
print(df6)

# Group the DataFrame by 'Category'
grouped = df6.groupby('Category')
print("\nDataFrame after group by column:")
print(grouped)

# compute the sum and mean of 'Value' for each category.
result=grouped.agg(['sum','mean'])
print(result)
```

```python
### Assignment 6: Time Series Analysis

# 1. Create a Pandas DataFrame with a datetime index and one column filled with random integers.
# Resample the DataFrame to compute the monthly mean of the values.

# Pandas DataFrame with a datetime index
data8={
     'Date': pd.date_range('2025-01-01', periods=90, freq='D'),
    'values': range(1, 91)#one column filled with random integers.
}
df8=pd.DataFrame(data8,columns=['Date','values'])
print(f"original dataframe")
print(df8)

# Set Date as index
df8.set_index('Date', inplace=True)

# Resample the DataFrame to compute the monthly mean of the values.
monthly_mean = df8.resample('ME').mean()

print(monthly_mean)
```

```python
# 2. Create a Pandas DataFrame with a datetime index ranging from '2021-01-01' to '2021-12-31' and one column filled with random integers. Compute the rolling mean with a window of 7 days.
import pandas as pd
import numpy as np

# Create a datetime index
dates = pd.date_range(start='2021-01-01', end='2021-12-31', freq='D')

# Create DataFrame with random integers
df = pd.DataFrame({
    'Value': np.random.randint(1, 100, size=len(dates))
}, index=dates)

# Compute 7-day rolling mean
df['Rolling_Mean_7D'] = df['Value'].rolling(window=7).mean()

print(df.head(10))
```

```python
### Assignment 7: MultiIndex DataFrame

# 1. Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.
import pandas as pd

# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [('A', 1),
     ('A', 2),
     ('A', 3),
     ('B', 1),
     ('B', 2),
     ('B', 3)],
    names=['Group', 'ID']
)

# Create DataFrame
df = pd.DataFrame(
    {'Value': [10, 20, 30, 40, 50, 60]},
    index=index
)

print(df)
#Access a specific row ('A', 2)
print(df.loc['A'])

#Basic Slicing Operations
# Slice all rows from Group 'A' to Group 'B'
print(df.loc['A':'B'])

#Select IDs 2 and 3 from all groups
print(df.loc[(slice(None), [2, 3]), :])
```

```python
# 2. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.
import pandas as pd
import numpy as np

# Create MultiIndex
index = pd.MultiIndex.from_tuples(
    [('Electronics', 'Mobile'),
     ('Electronics', 'Laptop'),
     ('Electronics', 'Tablet'),
     ('Furniture', 'Chair'),
     ('Furniture', 'Table'),
     ('Furniture', 'Sofa')],
    names=['Category', 'SubCategory']
)

# Create DataFrame with random values
df = pd.DataFrame(
    {'Value': np.random.randint(10, 100, size=len(index))},
    index=index
)

print("Original DataFrame:")
print(df)

# Sum for each Category
category_sum = df.groupby(level='Category').sum()

print("\nSum by Category:")
print(category_sum)

# Sum for each Category and SubCategory
subcategory_sum = df.groupby(level=['Category', 'SubCategory']).sum()

print("\nSum by Category and SubCategory:")
print(subcategory_sum)
```

```python
### Assignment 8: Pivot Tables
# 1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'.
# Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.
data_pivot={
    'Date':pd.date_range('2025-01-01', periods=90, freq='D'),
    'category':np.resize(['Electronics','Furniture','Kitchen','Home Decor','Clothes'],90),
    'value':np.random.randint(1,10, size=90)
}
data_pi=pd.DataFrame(data_pivot)
print(data_pi)

# Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.
result=data_pi.pivot_table(index='Date', columns='category', values='value',aggfunc='sum',fill_value=0)
print(result)
# result = data_pi.groupby(['Date', 'category'])['value'].sum().unstack(fill_value=0)
print(result)
```

```python
# 2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    'Year': [2022, 2022, 2022, 2022,
             2023, 2023, 2023, 2023,
             2024, 2024, 2024, 2024],
    'Quarter': ['Q1', 'Q2', 'Q3', 'Q4',
                'Q1', 'Q2', 'Q3', 'Q4',
                'Q1', 'Q2', 'Q3', 'Q4'],
    'Revenue': np.random.randint(1000, 5000, size=12)
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Create Pivot Table
pivot_table = df.pivot_table(
    values='Revenue',
    index='Quarter',
    columns='Year',
    aggfunc='mean'
)

print("\nMean Revenue by Quarter and Year:")
print(pivot_table)
```
