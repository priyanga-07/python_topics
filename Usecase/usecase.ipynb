```python
##### Example 1. Manage A To Do List
# - Create a To Do List To Keep Track OF Tasks

#creation of the list
to_do_list=[]
n=input("Enter the number of to do list:")
for i in range (int(n)):
    x=input("Enter the to do list:")
    to_do_list.append(x)
    print(to_do_list[i])

print("--------Do Anything Needed to be added to the list---------")

#More Task Has been added
status=input("Is there Anything Yet to be added the List(Yes/No)")
match status:
    case "Yes":
        xy=input("Enter the remaining to do list")
        to_do_list.append(xy)
        print(to_do_list)
    case "No":
        print("Thank You For The Day. Complete Today List Which is mentioned Below.")

print(f"Thank You For The Day. Complete Today List Which is mentioned Below. {to_do_list}")

#The task removed
x=input("Enter the task has been completed for the day:")
if x in to_do_list:
    to_do_list.remove(x)
    print(f"The Remaining Task in To Do List {to_do_list} ")
else:
    print("Task Not Found")
```

##### Example 2: Organizing Student Grades
- Create a list to store and calculate average grades for students

```python
score=[]
num=int(input("Enter the number of score:"))
for i in range(num):
    x=int(input("Enter the score:"))
    score.append(x)
    print({score[i]})

state=input("Is there anything Yet to be added the List(Yes/No)")
match state:
    case "Yes":
        xy=int(input("Enter the score"))
        score.append(xy)
        print(score)
    case "No":
        print("Thank You")
#renoval
remo=int(input("Enter the number to be removed:"))
if remo in score:
    score.remove(remo)
    print(f"The balance score LIst: {score}")
else:
    print("The balance score NOT FOUND")
#summation
total=sum(score)/len(score)
print(f"The total score is: {total}")
print(f"The maximum score is: {max(score)}")
print(f"The minimum score is: {min(score)}")
```

```python
##### Example 3: Managing An Inventory
# - Use a list to manage inventory items in a store

#Creation of inventory list

inv=[]
num=int(input("Enter the number of items:"))
for i in range(num):
    items=input("Enter the item:")
    inv.append(items)
    print(f"The item {items} is: {inv[i]}")

#Add/Remove
x=input("Do you want to Remove?(Add/Removal):")
y=input("Enter the item to be add/removal in the list")
match x:
    case "Removal":
        inv.remove(y)
        print(f"The item is: {inv}")
    case "Add":
        if y in inv:
            print(f"The item is already in the list {inv}")
        else:
            inv.append(y)
        print(f"The item is: {inv}")

# Printing the inventory
print("Inventory List:")
for item in inv:
    print(f"- {item}")

```

```python
##### Example 4:Collecting User Feedback
# - Use a list to collect and analyze user feedback.
feedback=[]
num=int(input("Enter the number of feedback:"))
for i in range(num):
    ls=input("Enter the feedback to be added:")
    feedback.append(ls)
    print(f"The feedback is: {feedback[i]}")

y=input("Can I continue to update the feedback (Add/Removal)")
match y:
    case "Add":
        f=input("Enter the feedback to be added:")
        feedback.append(f)
    case "Removal":
        f=input("Enter the feedback to be removed:")
        feedback.remove(f)
print(f"Thank You for the feedback: {feedback}")



# if y in feedback:
#     print(f"The feedback {y} is already in the list ")
# else:
#     feedback.append(y)
#     print(f"The feedback is added to the list: {feedback}")
print(f"Feedback List:")
for i in feedback:
    print(f"- {i}")

pos_count_feedback=sum(1 for comment in feedback if "good" in comment.lower() or "average" in comment.lower())
print(f"The total feedback is: {pos_count_feedback}")

neg_count_feedback=sum(1 for comment in feedback if "bad" in comment.lower() or "better" in comment.lower())
print(f"The total feedback is: {neg_count_feedback}")
```
