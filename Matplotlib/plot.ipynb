```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from unicodedata import category

df=pd.read_csv('student-dataset.csv')
df.head()

```

```python
#line plot
plt.plot(df['age'],df['gender'],linestyle='--',marker='o')
plt.xlabel('age')
plt.ylabel('gender')
plt.show()
```

```python
#multiple plot
plt.figure(figsize=(10,5))
plt.subplot(2,1,2)
plt.plot(df['age'],df['gender'],df['city'],linestyle='--',marker='o')
plt.title('Students data')

```

```python
plt.bar(df['age'],df['gender'])
plt.show()
```

```python
#histiogram
plt.hist(df['age'],bins=20)
plt.show()
```

```python
#scatter plot
plt.scatter(df['age'],df['english.grade'])
plt.show()
```
