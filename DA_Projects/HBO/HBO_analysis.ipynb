Most Common Genre

```python
#Chart 2:-Genre vs Release Year

# Splitting the Genre column
genre_split = hbo["Genre"].str.split(",")
# print(genre_split)
# Expanding the genre columns into separate rows
genre_split = genre_split.explode()
# Counting the genres
counting_genre = genre_split.value_counts()
# print(counting_genre)


# Plotting the top 10 genres
top_ten_genres = counting_genre.head(10)
ax = top_ten_genres.plot(kind="bar", figsize=(12, 6))

for bar in ax.patches:
    ax.annotate(
        f'{int(bar.get_height())}',
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha='center',
        va='bottom',
        fontsize=9
    )
plt.title("Top 10 Most Common Genre", fontsize=16)
plt.xlabel("Genre", fontsize=12)
plt.ylabel("Number of Titles", fontsize=12)
plt.grid(alpha=0.4)
plt.tight_layout()
plt.xticks(rotation=0)
plt.show()

```

Outlier Detection RT Score
The IQR method was used to detect outliers in the RT Score.

```python
#chart 8:-Boxplot
Q1 = hbo["RT Score"].quantile(0.25)
Q3 = hbo["RT Score"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

RTScore_outliers = hbo[(hbo["RT Score"]<lower_bound) | (hbo["RT Score"]>upper_bound)]
print(f"Number of Outliers of RT Score: {len(RTScore_outliers)}")

# Visualization
plt.figure(figsize=(8,4))
plt.boxplot(hbo["RT Score"],vert=False)
plt.xscale("log")
plt.title("Boxplot of Number of Rt score (Log Scale)")
plt.xlabel("Number of rt score (Log Scale)")
plt.show()
```

Outlier Detection-IMDb
The distribution of IMDb was examined using a boxplot and applied IQR method to detect potential outliers.


```python
#chart 9:-Box plot
Q1 = hbo["IMDb"].quantile(0.25)
Q3 = hbo["IMDb"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

IMDB_outliers = hbo[(hbo["IMDb"]<lower_bound) | (hbo["IMDb"]>upper_bound)]
print(f"Number of Outliers of IMDB: {len(IMDB_outliers)}")

# Visualization
plt.figure(figsize=(8,4))
plt.boxplot(hbo["IMDb"],'horizontal',vert=False)
plt.xscale("log")
plt.title("Boxplot of Number of IMDB (Log Scale)")
plt.xlabel("Number of IMDB (Log Scale)")
plt.show()
```
