Step 4:Visualization

```python
import matplotlib.pyplot as plt
#1.Avg monthly charges
plt.hist(df['MonthlyCharges'], bins=30)
plt.xlabel('MonthlyCharges')
plt.ylabel('Count')
plt.title('Distribution of Tele customer Churn')
plt.show()
```

```python
#2.AVg monthly charges by payment methods
df.groupby('PaymentMethod')['MonthlyCharges'].mean().plot(kind='bar',color='violet')
plt.xlabel('PaymentMethod')
plt.ylabel('Monthly Charges')
plt.title('Monthly Charges vs Payment Method')
plt.show()
```

```python
#3.avg monthly charge by payment and contract
df.groupby(['PaymentMethod','Contract'])['MonthlyCharges'].mean().unstack().plot(kind='bar',color=('r','g','b'))
plt.xlabel('PaymentMethod')
plt.ylabel('Monthly Charges')
plt.title('Monthly Charges by payment method and Contract')
plt.show()
```

```python
df.groupby('tenure')['MonthlyCharges'].mean().plot(kind='line')
#4.avg monthly charge by tenure
plt.title('Average Monthly Charges by Tenure')
plt.xlabel('Tenure')
plt.ylabel('Average Monthly Charges')
plt.show()
```

```python
#5.avg monthly charge by payment,contract,internet service
df.groupby(['PaymentMethod','Contract','InternetService'])['MonthlyCharges'].mean().unstack().plot(kind='bar',color=('r','g','b'))
# plt.figure(figsize=(12, 8))
plt.xlabel('PaymentMethod')
plt.ylabel('Monthly Charges')
plt.title('Monthly Charges by Payment Method,contract and InternetService')
plt.show()
```

```python
#6.pie chart
df['Contract'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    ylabel=''
)

plt.title('Customers by Contract Type')
plt.show()
```

```python
    #7.avg monthly charge by tenure and churn
df.groupby(['Churn'])[['MonthlyCharges','tenure']].mean().plot(kind='barh')
plt.xlabel('Churn')
plt.ylabel('MonthlyCharges')
plt.title('Monthly Charges by Churn')
plt.show()
```

```python
plt.boxplot(df['MonthlyCharges'])
# 8.boxplot
plt.title('Monthly Charges Box Plot')
plt.show()
```

```python
plt.scatter(df['tenure'], df['MonthlyCharges'])

plt.title('Tenure vs Monthly Charges')
plt.xlabel('Tenure')
plt.ylabel('Monthly Charges')
plt.show()
```
