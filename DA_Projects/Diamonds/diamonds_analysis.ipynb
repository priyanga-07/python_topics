```python
# Summary statistics
print(df.describe())
```

Step 2: Data Cleaning
Check for missing values:

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates() #Remove duplicates (if any):
df
```

Step 3: Exploratory Data Analysis (EDA)

```python
print(df['price'].mean()) #What is the average price of diamonds?
```

```python
print(df.groupby('cut')['price'].mean()) #Which cut has the highest average price?
```

```python
print(df.groupby('color')['price'].mean()) #Average price by color
```

```python
print(df.groupby('clarity')['price'].mean()) #Average price by clarity
```

Step 4: Visualizations

```python
import matplotlib.pyplot as plt
#Price Distribution
plt.hist(df['price'], bins=30)
plt.xlabel('Price')
plt.ylabel('Count')
plt.title('Distribution of Diamond Prices')
plt.show()
```

```python
df.groupby('cut')['price'].mean().plot(kind='bar')#Average Price by Cut
plt.ylabel('Average Price')
plt.show()
```

```python
#Carat vs Price
plt.scatter(df['carat'], df['price'])
plt.xlabel('Carat')
plt.ylabel('Price')
plt.title('Carat vs Price')
plt.show()
```

```python
print(df.corr(numeric_only=True))#Step 5: Advanced Analysis
#Correlation between carat, depth, table, x, y, z, and price.
```
